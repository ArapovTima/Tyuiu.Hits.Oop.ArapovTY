```
public class Plant
{
    public string Name{get; set;}
    public string Type{get; set;}
    public float Height{get; set;}

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine(Name + " начинает цветение");
    }

    public void IncreaseSize(float growthAmount)
    {
        Height += growthAmount;
        Console.WriteLine(Name + " прибавил в росте " + growthAmount + " см. Общая высота теперь: " + Height + " см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalsCount { get; set; }

    public Flower(string name, float height, string color, int petalsCount) 
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalsCount = petalsCount;
    }

    public override void Bloom()
    {
        Console.WriteLine("Бутоны " + Name + " раскрываются и создают прекрасный вид");
    }

    public void StartWilting()
    {
        Console.WriteLine(Name + " постепенно теряет свою свежесть");
    }
}

public class Tree : Plant
{
    public int Age { get; set; } 
    public string LeavesType { get; set; }

    public Tree(string name, float height, int age, string leavesType) 
        : base(name, "Дерево", height)
    {
        Age = age; 
        LeavesType = leavesType; 
    }

    public override void Bloom()
    {
        Console.WriteLine("На " + Name + " появляется молодая листва");
    }

    public void AddYears(int yearsToAdd) 
    {
        Age += yearsToAdd;
        Console.WriteLine(Name + " стал более зрелым на " + yearsToAdd + " год/года. Текущий возраст: " + Age + " лет");
    }
}
        
        
Flower myFlower = new Flower("Роза", 50, "Красный", 12);       
Tree myTree = new Tree("Дуб", 300, 10, "Лиственные");

Console.WriteLine("Мой цветок:");
Console.WriteLine("Имя: " + myFlower.Name);
Console.WriteLine("Цвет: " + myFlower.Color);
Console.WriteLine("Лепестков: " + myFlower.PetalsCount);
        
myFlower.Bloom();
myFlower.IncreaseSize(15);
myFlower.StartWilting();
Console.WriteLine();
Console.WriteLine("Мое дерево:");
Console.WriteLine("Имя: " + myTree.Name);
Console.WriteLine("Тип листьев: " + myTree.LeavesType);
Console.WriteLine("Возраст: " + myTree.Age + " лет");
        
myTree.Bloom();
myTree.IncreaseSize(25);
myTree.AddYears(2);
Console.WriteLine();
```
