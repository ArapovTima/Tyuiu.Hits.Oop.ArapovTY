### Вариант задания 2


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс для всех типов студентов
public class Student
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }
    
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
    }
    
    // Виртуальные методы для переопределения в наследниках
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}";
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }
}

// Студент бакалавриата - добавляет специальность
public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    
    public BachelorStudent(string name, int age, int course, string specialty) 
        : base(name, age, course)
    {
        Specialty = specialty;
    }
    
    // Переопределяет метод получения информации
    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty} (Бакалавриат)";
    }
    
    // Переопределяет метод обучения с учетом специальности
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности '{Specialty}'");
    }
}

// Студент магистратуры - добавляет научного руководителя
public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }
    
    public MasterStudent(string name, int age, int course, string scientificAdvisor) 
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
    }
    
    // Переопределяет метод получения информации
    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor} (Магистратура)";
    }
    
    // Переопределяет метод сдачи экзамена с участием руководителя
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }
}

// Аспирант - добавляет тему диссертации
public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }
    
    public PhDStudent(string name, int age, int course, string dissertationTopic) 
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
    }
    
    // Переопределяет метод получения информации
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: '{DissertationTopic}' (Аспирантура)";
    }
    
    // Переопределяет метод обучения как исследовательскую работу
    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему: '{DissertationTopic}'");
    }
}

// Основной класс программы
class Program
{
    public static void Main()
    {
        // Демонстрация полиморфизма: массив базового типа содержит объекты разных типов
        Student[] students = {
            new BachelorStudent("Дмитрий Соколов", 20, 2, "Информатика"),
            new MasterStudent("Светлана Иванова", 22, 3, "д.т.н. Волков"),
            new PhDStudent("Александр Петров", 23, 4, "Искусственный интеллект")
        };
        
        // Для каждого студента вызываются переопределенные методы
        foreach (var student in students)
        {
            Console.WriteLine(student.GetInfo());  // Полиморфный вызов GetInfo()
            student.Study();                       // Полиморфный вызов Study()
            student.TakeExam();                    // Полиморфный вызов TakeExam()
            Console.WriteLine();
        }
    }
}

// Запуск программы
Program.Main();
```
