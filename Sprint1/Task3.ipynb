### Вариант задания 2


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Student
{
    private string _name;
    private int _age;
    private int _course;
    
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 16 && value <= 70)
                _age = value;
        }
    }
    
    public int Course
    {
        get { return _course; }
        set 
        { 
            if (value >= 1 && value <= 6)
                _course = value;
        }
    }
    
    public static int TotalStudents { get; private set; } = 0;
    
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        TotalStudents++;
    }
    
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}";
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }
    
    public void InteractWith(Student otherStudent)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherStudent.Name}");
    }
    
    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Общее количество студентов: {TotalStudents}");
    }
}

public class BachelorStudent : Student
{
    private string _specialty;
    
    public string Specialty
    {
        get { return _specialty; }
        set { _specialty = value; }
    }
    
    public BachelorStudent(string name, int age, int course, string specialty) 
        : base(name, age, course)
    {
        Specialty = specialty;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности '{Specialty}'");
    }
    
    public void DoLabWork()
    {
        Console.WriteLine($"{Name} выполняет лабораторную работу");
    }
}

public class MasterStudent : Student
{
    private string _scientificAdvisor;
    
    public string ScientificAdvisor
    {
        get { return _scientificAdvisor; }
        set { _scientificAdvisor = value; }
    }
    
    public MasterStudent(string name, int age, int course, string scientificAdvisor) 
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}";
    }
    
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }
    
    public void WriteThesis()
    {
        Console.WriteLine($"{Name} пишет диссертацию");
    }
}

public class PhDStudent : Student
{
    private string _dissertationTopic;
    
    public string DissertationTopic
    {
        get { return _dissertationTopic; }
        set { _dissertationTopic = value; }
    }
    
    public PhDStudent(string name, int age, int course, string dissertationTopic) 
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: '{DissertationTopic}'";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему: '{DissertationTopic}'");
    }
    
    public void DefendDissertation()
    {
        Console.WriteLine($"{Name} защищает диссертацию");
    }
}

BachelorStudent bachelor = new BachelorStudent("Дмитрий Соколов", 20, 2, "Информатика");
MasterStudent master = new MasterStudent("Светлана Иванова", 22, 3, "д.т.н. Волков");
PhDStudent phd = new PhDStudent("Александр Петров", 23, 4, "Искусственный интеллект");

Console.WriteLine(bachelor.GetInfo());
Console.WriteLine(master.GetInfo());
Console.WriteLine(phd.GetInfo());

bachelor.Study();
master.Study();
phd.Study();

bachelor.TakeExam();
master.TakeExam();
phd.TakeExam();

bachelor.InteractWith(master);
master.InteractWith(phd);
phd.InteractWith(bachelor);

bachelor.DoLabWork();
master.WriteThesis();
phd.DefendDissertation();

Student.DisplayTotalStudents();
```
