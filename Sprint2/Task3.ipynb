Базовый класс Student

### Вариант задания 2


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый интерфейс для всех студентов
public interface IStudent
{
    string Name { get; set; }
    int Age { get; set; }
    void Study();
    void TakeExam();
}

// Интерфейс для научной деятельности
public interface IResearch
{
    string ResearchTopic { get; set; }
    void ConductResearch();
}

// Интерфейс для практической деятельности
public interface IPractical
{
    string InternshipCompany { get; set; }
    void DoInternship();
}

// Базовый класс Student
public class Student : IStudent
{
    private string _name;
    private int _age;
    private int _course;
    
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 16 && value <= 70)
                _age = value;
        }
    }
    
    public int Course
    {
        get { return _course; }
        set 
        { 
            if (value >= 1 && value <= 6)
                _course = value;
        }
    }
    
    public string Email { get; set; }
    public double GPA { get; set; }
    
    public static int TotalStudents { get; private set; } = 0;
    
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        TotalStudents++;
    }
    
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}, GPA: {GPA}";
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }
    
    public void InteractWith(Student otherStudent)
    {
        Console.WriteLine($"{Name} взаимодействует с {otherStudent.Name}");
    }
    
    public void UpdateEmail(string email)
    {
        Email = email;
        Console.WriteLine($"{Name} обновил email на: {email}");
    }
    
    public void CalculateGPA(double[] grades)
    {
        double sum = 0;
        foreach (double grade in grades)
            sum += grade;
        GPA = sum / grades.Length;
        Console.WriteLine($"{Name} имеет GPA: {GPA:F2}");
    }
    
    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Общее количество студентов: {TotalStudents}");
    }
}

// Простое наследование - BachelorStudent наследует только от Student
public class BachelorStudent : Student
{
    private string _specialty;
    
    public string Specialty
    {
        get { return _specialty; }
        set { _specialty = value; }
    }
    
    public string Faculty { get; set; }
    public int Scholarship { get; set; }
    
    public BachelorStudent(string name, int age, int course, string specialty) 
        : base(name, age, course)
    {
        Specialty = specialty;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}, Факультет: {Faculty}";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности '{Specialty}' на факультете {Faculty}");
    }
    
    public void DoLabWork()
    {
        Console.WriteLine($"{Name} выполняет лабораторную работу по {Specialty}");
    }
    
    public void ApplyForScholarship()
    {
        Console.WriteLine($"{Name} подает заявку на стипендию");
    }
}

// Сложное наследование - MasterStudent наследует от BachelorStudent
public class MasterStudent : BachelorStudent, IResearch
{
    private string _scientificAdvisor;
    
    public string ScientificAdvisor
    {
        get { return _scientificAdvisor; }
        set { _scientificAdvisor = value; }
    }
    
    public string ResearchTopic { get; set; }
    public string ThesisTitle { get; set; }
    
    public MasterStudent(string name, int age, int course, string specialty, string scientificAdvisor) 
        : base(name, age, course, specialty)
    {
        ScientificAdvisor = scientificAdvisor;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}, Тема исследования: {ResearchTopic}";
    }
    
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }
    
    public void WriteThesis()
    {
        Console.WriteLine($"{Name} пишет диссертацию: '{ThesisTitle}'");
    }
    
    public void ConductResearch()
    {
        Console.WriteLine($"{Name} проводит исследование по теме: '{ResearchTopic}'");
    }
    
    public void PublishArticle()
    {
        Console.WriteLine($"{Name} публикует научную статью");
    }
}

// Множественное наследование через интерфейсы - PhDStudent наследует от MasterStudent и реализует дополнительные интерфейсы
public class PhDStudent : MasterStudent, IResearch, IPractical
{
    private string _dissertationTopic;
    
    public string DissertationTopic
    {
        get { return _dissertationTopic; }
        set { _dissertationTopic = value; }
    }
    
    public string InternshipCompany { get; set; }
    public int PublicationsCount { get; set; }
    public string ResearchGroup { get; set; }
    
    public PhDStudent(string name, int age, int course, string specialty, string scientificAdvisor, string dissertationTopic) 
        : base(name, age, course, specialty, scientificAdvisor)
    {
        DissertationTopic = dissertationTopic;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: '{DissertationTopic}', Публикации: {PublicationsCount}";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему: '{DissertationTopic}' в исследовательской группе {ResearchGroup}");
    }
    
    public void DefendDissertation()
    {
        Console.WriteLine($"{Name} защищает диссертацию на тему: '{DissertationTopic}'");
    }
    
    public void DoInternship()
    {
        Console.WriteLine($"{Name} проходит стажировку в компании {InternshipCompany}");
    }
    
    public void AttendConference()
    {
        Console.WriteLine($"{Name} участвует в научной конференции");
    }
}

BachelorStudent bachelor = new BachelorStudent("Дмитрий Соколов", 20, 2, "Информатика");
MasterStudent master = new MasterStudent("Светлана Иванова", 22, 3, "Компьютерные науки", "д.т.н. Волков");
PhDStudent phd = new PhDStudent("Александр Петров", 23, 4, "Искусственный интеллект", "д.ф.-м.н. Сидоров", "Нейросети в медицине");

bachelor.Faculty = "Факультет информационных технологий";
bachelor.Scholarship = 1500;

master.ResearchTopic = "Машинное обучение";
master.ThesisTitle = "Алгоритмы глубокого обучения";

phd.ResearchGroup = "AI Research Lab";
phd.InternshipCompany = "Google Research";
phd.PublicationsCount = 5;

bachelor.CalculateGPA(new double[] { 4.5, 4.0, 4.2, 4.7 });
master.CalculateGPA(new double[] { 4.8, 4.9, 4.7, 5.0 });
phd.CalculateGPA(new double[] { 5.0, 5.0, 4.9, 5.0 });

bachelor.UpdateEmail("dmitry.sokolov@edu.ru");
master.UpdateEmail("svetlana.ivanova@edu.ru");
phd.UpdateEmail("alexander.petrov@edu.ru");

Console.WriteLine(bachelor.GetInfo());
Console.WriteLine(master.GetInfo());
Console.WriteLine(phd.GetInfo());

bachelor.Study();
master.Study();
phd.Study();

bachelor.TakeExam();
master.TakeExam();
phd.TakeExam();

bachelor.DoLabWork();
bachelor.ApplyForScholarship();

master.WriteThesis();
master.ConductResearch();
master.PublishArticle();

phd.DefendDissertation();
phd.DoInternship();
phd.AttendConference();
phd.ConductResearch();

bachelor.InteractWith(master);
master.InteractWith(phd);

Student.DisplayTotalStudents();
```
