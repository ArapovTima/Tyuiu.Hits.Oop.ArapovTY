```
using System;
using System.Collections.Generic;

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; }
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; }
}

public class Enterprise : IEnterprise
{
    private List<ISupplier> suppliers = new List<ISupplier>();
    private List<IEmployee> employees = new List<IEmployee>();

    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        return suppliers;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal total = 0;
        
        foreach (var supplier in suppliers)
        {
            total += supplier.GetCost();
        }
        
        foreach (var employee in employees)
        {
            total += employee.GetSalary();
        }
        
        return total;
    }
}

public class LogisticsSupplier : ISupplier
{
    private string name;
    private decimal cost;

    public string SupplierType { get; } = "Логистика";

    public LogisticsSupplier(string name, decimal cost)
    {
        this.name = name;
        this.cost = cost;
    }

    public string GetName()
    {
        return name;
    }

    public decimal GetCost()
    {
        return cost;
    }
}

public class EquipmentSupplier : ISupplier
{
    private string name;
    private decimal cost;

    public string SupplierType { get; } = "Офисное оборудование";

    public EquipmentSupplier(string name, decimal cost)
    {
        this.name = name;
        this.cost = cost;
    }

    public string GetName()
    {
        return name;
    }

    public decimal GetCost()
    {
        return cost;
    }
}

public class MarketingSupplier : ISupplier
{
    private string name;
    private decimal cost;

    public string SupplierType { get; } = "Маркетинг";

    public MarketingSupplier(string name, decimal cost)
    {
        this.name = name;
        this.cost = cost;
    }

    public string GetName()
    {
        return name;
    }

    public decimal GetCost()
    {
        return cost;
    }
}

public class Manager : IEmployee
{
    private string fullName;
    private decimal salary;

    public string Position { get; } = "Менеджер";

    public Manager(string fullName, decimal salary)
    {
        this.fullName = fullName;
        this.salary = salary;
    }

    public string GetFullName()
    {
        return fullName;
    }

    public decimal GetSalary()
    {
        return salary;
    }
}

public class Developer : IEmployee
{
    private string fullName;
    private decimal salary;

    public string Position { get; } = "Разработчик";

    public Developer(string fullName, decimal salary)
    {
        this.fullName = fullName;
        this.salary = salary;
    }

    public string GetFullName()
    {
        return fullName;
    }

    public decimal GetSalary()
    {
        return salary;
    }
}

public class Designer : IEmployee
{
    private string fullName;
    private decimal salary;

    public string Position { get; } = "Дизайнер";

    public Designer(string fullName, decimal salary)
    {
        this.fullName = fullName;
        this.salary = salary;
    }

    public string GetFullName()
    {
        return fullName;
    }

    public decimal GetSalary()
    {
        return salary;
    }
}

public class Accountant : IEmployee
{
    private string fullName;
    private decimal salary;

    public string Position { get; } = "Бухгалтер";

    public Accountant(string fullName, decimal salary)
    {
        this.fullName = fullName;
        this.salary = salary;
    }

    public string GetFullName()
    {
        return fullName;
    }

    public decimal GetSalary()
    {
        return salary;
    }
}

Enterprise enterprise = new Enterprise();

ISupplier supplier1 = new LogisticsSupplier("Быстрая доставка", 120000);
ISupplier supplier2 = new EquipmentSupplier("ОфисТехника", 75000);
ISupplier supplier3 = new MarketingSupplier("Рекламное агентство", 90000);

IEmployee employee1 = new Manager("Петр Сидоров", 80000);
IEmployee employee2 = new Developer("Мария Козлова", 110000);
IEmployee employee3 = new Designer("Алексей Волков", 85000);
IEmployee employee4 = new Accountant("Елена Смирнова", 70000);

enterprise.AddSupplier(supplier1);
enterprise.AddSupplier(supplier2);
enterprise.AddSupplier(supplier3);

enterprise.AddEmployee(employee1);
enterprise.AddEmployee(employee2);
enterprise.AddEmployee(employee3);
enterprise.AddEmployee(employee4);

Console.WriteLine("Список поставщиков:");
var suppliers = enterprise.GetSuppliers();
for (int i = 0; i < suppliers.Count; i++)
{
    Console.WriteLine($"{i + 1}. {suppliers[i].GetName()} – Стоимость услуг: {suppliers[i].GetCost()}");
}

Console.WriteLine("\nСписок сотрудников:");
Console.WriteLine($"1. {employee1.GetFullName()} – Должность: {employee1.Position} – Зарплата: {employee1.GetSalary()}");
Console.WriteLine($"2. {employee2.GetFullName()} – Должность: {employee2.Position} – Зарплата: {employee2.GetSalary()}");
Console.WriteLine($"3. {employee3.GetFullName()} – Должность: {employee3.Position} – Зарплата: {employee3.GetSalary()}");
Console.WriteLine($"4. {employee4.GetFullName()} – Должность: {employee4.Position} – Зарплата: {employee4.GetSalary()}");

Console.WriteLine($"\nОбщие расходы: {enterprise.CalculateTotalExpenses()}");
```
