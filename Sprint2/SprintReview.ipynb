```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set 
        { 
            if (value >= 0)
                price = value;
        }
    }

    public int Quantity
    {
        get { return quantity; }
        set 
        { 
            if (value >= 0)
                quantity = value;
        }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        if (quantity + amount >= 0)
            quantity += amount;
    }

    public string GetProductInfo()
    {
        return name + " - " + price + " руб, " + quantity + " шт";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set 
        { 
            if (value >= 0)
                salary = value;
        }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine(name + " добавил товар " + product.Name);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (products.Contains(product))
        {
            if (product.Quantity >= quantity)
            {
                product.UpdateQuantity(-quantity);
                Console.WriteLine(name + " продал " + quantity + " шт " + product.Name);
            }
            else
            {
                Console.WriteLine("Мало товара " + product.Name);
            }
        }
        else
        {
            Console.WriteLine(name + " не продает " + product.Name);
        }
    }

    public string GetSellerInfo()
    {
        return name + " (ID: " + employeeId + "), зарплата: " + salary + " руб";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();
    private List<Product> products = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine(seller.Name + " работает в " + storeName);
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine(product.Name + " есть в " + storeName);
    }

    public void ListProducts()
    {
        Console.WriteLine("Товары в " + storeName + ":");
        foreach (var product in products)
        {
            Console.WriteLine(product.GetProductInfo());
        }
    }

    public string GetStoreInfo()
    {
        return storeName + " (" + location + "), работает: " + storeHours;
    }
}

Product laptop = new Product("Ноутбук", 50000, 10, "Техника", "Хороший ноут");
Product phone = new Product("Телефон", 30000, 5, "Техника", "Новый телефон");
Product book = new Product("Книга", 500, 20, "Книги", "Интересная книга");

Seller seller1 = new Seller("Иван", "001", 40000, "ivan@mail.ru");
Seller seller2 = new Seller("Мария", "002", 45000, "maria@mail.ru");

Store store = new Store("ТехноМир", "ул. Сибирская 15", "8-20");

store.AddProduct(laptop);
store.AddProduct(phone);
store.AddProduct(book);

store.AddSeller(seller1);
store.AddSeller(seller2);

seller1.AddProduct(laptop);
seller1.AddProduct(phone);
seller2.AddProduct(book);

Console.WriteLine(store.GetStoreInfo());

Console.WriteLine(seller1.GetSellerInfo());
Console.WriteLine(seller2.GetSellerInfo());

Console.WriteLine("До продажи:");
store.ListProducts();

seller1.SellProduct(laptop, 2);
seller2.SellProduct(book, 3);
seller1.SellProduct(phone, 10);

Console.WriteLine("После продажи:");
store.ListProducts();

Console.WriteLine("Всего ноутбуков на: " + laptop.GetTotalPrice() + " руб");
Console.WriteLine("Всего телефонов на: " + phone.GetTotalPrice() + " руб");
Console.WriteLine("Всего книг на: " + book.GetTotalPrice() + " руб");
```
